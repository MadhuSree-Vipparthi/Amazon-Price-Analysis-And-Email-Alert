```python
import requests
from bs4 import BeautifulSoup
import smtplib

# AMAZON PRODUCT URL
url = "https://www.amazon.in/ASUS-Vivobook-Metallic-Windows-S3407CA-LY057WS/dp/B0F9YGSK6L/"

# EMAIL CONFIGURATION
EMAIL = "Your_Mail"
APP_PASSWORD = "Enter_Your_App_Password"
TARGET_PRICE = 90000

# SEND EMAIL FUNCTION
def send_mail(title, price):
    print("\nConnecting to Gmail...")
    server = smtplib.SMTP_SSL("smtp.gmail.com", 465)

    print("Logging in...")
    server.login(EMAIL, APP_PASSWORD)

    subject = "Laptop Price Alert!"
    body = f"""
Good News!
Product:
{title}
Current Price: Rs.{price}
Target Price: Rs.{TARGET_PRICE}
Buy Now:
{url}
"""

    msg = f"Subject: {subject}\n\n{body}"
    print("Sending Email...")
    server.sendmail(
        EMAIL,
        EMAIL,
        msg
    )

    server.quit()

    print("Email Sent Successfully")

# SCRAPE AMAZON
headers = {
    "User-Agent":
    "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/78.0.3904.108 Safari/537.36",
    "Accept-Encoding":"gzip, deflate",
    "Accept":"text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8",
    "DNT":"1",
    "Connection":"close",
    "Upgrade-Insecure-Requests":"1"
}

print("Fetching Amazon page...")

response = requests.get(url, headers=headers)

soup1 = BeautifulSoup(response.content, "html.parser")
soup2 = BeautifulSoup(soup1.prettify(), "html.parser")


# PRODUCT NAME
title = soup2.find(id='productTitle').get_text()

# PRODUCT PRICE
price_text = soup2.find(class_='a-price-whole').get_text()
clean_price = ''.join(filter(str.isdigit, price_text))
current_price = int(clean_price)

# DISPLAY DETAILS
print("\nProduct Name:")
print(title)

print("\nCurrent Price:")
print(current_price)

# PRICE CHECK
if current_price <= TARGET_PRICE:

    print("\n Price is below target")

    send_mail(title, current_price)

else:

    print("\n Price is still above target")
    print("Target Price:", TARGET_PRICE)
```
